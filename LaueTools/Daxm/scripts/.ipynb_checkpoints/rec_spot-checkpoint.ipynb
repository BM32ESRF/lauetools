**Goal:   Find manually a depth-range `drange` (min, max) values for the scattering intensity reconstruction.**

Reconstruction will add a new dimension (depth along beam) in any requested X-ray detector pixel (i.e pixels in ROI centered on Laue spots). For any pixel having this extra dimension, the element is called voxel.

Since zero depth may be inaccurate (even after wire trajectories calibration step), and the computation time increases with the nb of voxels,  `drange` (min value, max value) (in mm) and `ystep` (in mm) must fit rather the region where reconstructed signal is non zero. (grain or grains location).

In this notebook, by choosing 1 Laue spots we may adapt the shorter `drange_narrow` allowing the reconstruction of voxel scattering intensity of the corresponding grain.  

To handle larger depth for the reconstruction (to get deeper grains) or several daxm scan measured at different sample height of the specimen with a tilted surface (e.g. about 40 degrees on BM32), a larger range will be necessary. In this notebook this is called `drange_large` and must be used afterwards as input parameter of reconstruction in rec.py script#

**ybeam positive towards sample depth (and along the beam propagation direction)**

recspot, depth = rec.interpolate(drange_narrow)

```python
# a helper function to visualise 3D data and with 2D slices
def fig_slider(vol, axis=2, datacontent='XYdepth'): # or YXdepth
    
    vol = np.array(vol)  # 3D data
    
    assert vol.ndim ==3
    
    #widget = widgets.IntSlider(min=0, max=vol.shape[axis]-1, step=1, value=0)
    
    mplp.figure(figsize=(5,3))
    
    if axis==0:
        img0 = vol[0, :, :]
    elif axis==1:
        img0 = vol[:, 0, :]
    else: # axis = 2
        img0 = vol[:, :, 0].T
    
    ax = mplp.imshow(np.flipud(img0), vmin=np.min(vol), vmax=np.mean(vol) + 3.*np.std(vol), origin="upper")
    mplp.colorbar()

    if axis==2:
        if datacontent == 'XYdepth':
            mplp.xlabel('detector pixel X')
            mplp.ylabel('detector pixel Y')
        mplp.xlabel('detector pixel X')
        mplp.ylabel('detector pixel Y')
        
    if axis==0:
        
        mplp.xlabel('depth index ')
        mplp.ylabel('detector pixel X')
        
    if axis==1:
        
        mplp.xlabel('depth index ')
        mplp.ylabel('detector pixel Y')
        
    def update_img(idx=0):
        if axis==0:
            img = vol[idx, :, :]
        elif axis==1:
            img = vol[:, idx, :]
        else:
            img = vol[:, :, idx].T
        if axis==2 and datacontent == 'XYdepth':
            mplp.title('pixel intensity (changing depth index=%d)'%idx)
        
        ax.set_data(img)
        mplp.gca().set_aspect('equal')
    
    #return interact(update_img, idx=widget)
    return interact(update_img, idx=(0, vol.shape[axis]-1))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.measure import marching_cubes
from mpl_toolkits.mplot3d import Axes3D
from ipywidgets import interact, FloatSlider, IntSlider

# Your 3D intensity array (shape: pixel X, pixel Y, depth)
intensity_3d = recspot  # Replace with your data

# Normalize intensity
intensity_normalized = (intensity_3d - intensity_3d.min()) / (intensity_3d.max() - intensity_3d.min())

# Create a figure and axis
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# Function to update the plot based on the threshold and depth range
def update_plot(threshold=0.5, depth_min=0, depth_max=59):
    ax.clear()  # Clear the previous plot

    # Slice the intensity array based on depth range (axis 2)
    sliced_intensity = intensity_normalized[:, :, depth_min:depth_max+1]

    # Compute isosurface for the sliced data
    verts, faces, _, _ = marching_cubes(sliced_intensity, level=threshold)

    # Adjust vertex coordinates to match the original depth range
    verts[:, 2] += depth_min  # Shift vertices to the correct depth position

    # Plot the isosurface
    mesh = ax.plot_trisurf(
        verts[:, 0], verts[:, 1], faces, verts[:, 2],
        cmap='viridis',
        lw=0.1,
        alpha=0.8,
        edgecolor='none'
    )

    # Label axes
    ax.set_xlabel('Pixel X')
    ax.set_ylabel('Pixel Y')
    ax.set_zlabel('Depth')
    ax.set_title(f'3D Isosurface (Threshold: {threshold:.2f}, Depth: {depth_min}-{depth_max})')

    # Set consistent axis limits to prevent the plot from shrinking
    ax.set_xlim([0, intensity_3d.shape[0]])
    ax.set_ylim([0, intensity_3d.shape[1]])
    ax.set_zlim([0, intensity_3d.shape[2]])

    # Redraw the figure
    fig.canvas.draw_idle()

# Initialize the plot with default values
update_plot(threshold=0.5, depth_min=0, depth_max=intensity_3d.shape[2]-1)

# Create interactive sliders
interact(
    update_plot,
    threshold=FloatSlider(min=0.0, max=1.0, step=0.05, value=0.5, description='Threshold:'),
    depth_min=IntSlider(min=0, max=intensity_3d.shape[2]-1, step=1, value=0, description='Depth Min:'),
    depth_max=IntSlider(min=0, max=intensity_3d.shape[2]-1, step=1, value=intensity_3d.shape[2]-1, description='Depth Max:')
)

#plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pyvista as pv

pv.set_jupyter_backend("trame")

# ==========================================================
# INPUT DATA
# ==========================================================

volume = recspot.astype(np.float32)

# Normalize
volume -= volume.min()
if volume.max() > 0:
    volume /= volume.max()

nx, ny, nz = volume.shape

# ==========================================================
# INITIAL PARAMETERS
# ==========================================================

threshold = 0.5
depth_min = 0
depth_max = nz - 1

# ==========================================================
# PLOTTER
# ==========================================================

plotter = pv.Plotter(window_size=(1200, 900))

surface_actor = None
slice_actor = None
outline_actor = None

# ==========================================================
# BUILD SCENE
# ==========================================================

def build_scene():

    global surface_actor, slice_actor, outline_actor

    # Remove previous actors
    if surface_actor is not None:
        plotter.remove_actor(surface_actor)

    if slice_actor is not None:
        plotter.remove_actor(slice_actor)

    if outline_actor is not None:
        plotter.remove_actor(outline_actor)

    # ------------------------------------------------------
    # Crop volume
    # ------------------------------------------------------

    cropped = volume[:, :, depth_min:depth_max+1]

    # ------------------------------------------------------
    # Create ImageData
    # ------------------------------------------------------

    grid = pv.ImageData()

    grid.dimensions = np.array(cropped.shape) + 1
    grid.spacing = (1, 1, 1)
    grid.origin = (0, 0, depth_min)

    grid.cell_data["values"] = cropped.flatten(order="F")

    # ------------------------------------------------------
    # Isosurface
    # ------------------------------------------------------

    try:

        contour = grid.contour(
            [threshold],
            scalars="values"
        )

        surface_actor = plotter.add_mesh(
            contour,
            cmap="viridis",
            opacity=0.85,
            smooth_shading=True,
            name="surface"
        )

    except Exception:
        pass

    # ------------------------------------------------------
    # Orthogonal slices
    # ------------------------------------------------------

    slices = grid.slice_orthogonal(
        x=nx//2,
        y=ny//2,
        z=(depth_max-depth_min)//2
    )

    slice_actor = plotter.add_mesh(
        slices,
        cmap="gray",
        opacity=0.35,
        clim=[0,1],
        name="slices"
    )

    # ------------------------------------------------------

    outline_actor = plotter.add_mesh(
        grid.outline(),
        color="black",
        line_width=2
    )

    plotter.render()


# ==========================================================
# CALLBACKS
# ==========================================================

def threshold_callback(value):

    global threshold
    threshold = value
    build_scene()


def depth_min_callback(value):

    global depth_min

    value = int(value)

    if value < depth_max:
        depth_min = value
        build_scene()


def depth_max_callback(value):

    global depth_max

    value = int(value)

    if value > depth_min:
        depth_max = value
        build_scene()


# ==========================================================
# INITIAL SCENE
# ==========================================================

build_scene()

# ==========================================================
# SLIDERS
# ==========================================================

plotter.add_slider_widget(
    threshold_callback,
    rng=[0,1],
    value=threshold,
    title="Threshold",
    pointa=(0.03,0.1),
    pointb=(0.30,0.1),
)

plotter.add_slider_widget(
    depth_min_callback,
    rng=[0,nz-2],
    value=0,
    title="Depth min",
    pointa=(0.03,0.18),
    pointb=(0.30,0.18),
)

plotter.add_slider_widget(
    depth_max_callback,
    rng=[1,nz-1],
    value=nz-1,
    title="Depth max",
    pointa=(0.03,0.26),
    pointb=(0.30,0.26),
)

# ==========================================================
# AXES
# ==========================================================

plotter.add_axes()

plotter.show_grid()

plotter.show()
```
