```python
import os
from glob import glob

# Base directory
base_dir = "/data/visitor"

# Pattern to match the directory structure and file names
pattern = os.path.join(
    base_dir,
    "a32*",  # Only folders starting with 'a32'
    "bm32",
    "*",
    "RAW_DATA",
    "*",
    "*",
    "*.h5"
)

# Find all matching .h5 files
h5_files = glob(pattern, recursive=True)

# Filter files containing 'align' or 'beamline' in their names
filtered_files = [
    file for file in h5_files
    if ('align' in os.path.basename(file).lower() or 'beamline' in os.path.basename(file).lower())
]

# Calculate accumulated size
accumulated_size = 0
file_info = []
for file in filtered_files:
    size = os.path.getsize(file)  # Size in bytes
    accumulated_size += size
    file_info.append((file, size))

# Print individual file sizes (optional)
for file, size in file_info:
    print(f"File: {file}\nSize: {size} bytes\n---")

# Print accumulated size
print(f"\nAccumulated size of all files: {accumulated_size} bytes")
print(f"Accumulated size in MB: {accumulated_size / (1024 * 1024):.2f} MB")
print(f"Accumulated size in GB: {accumulated_size / (1024 * 1024 * 1024):.2f} GB")
```

```python
import os
from glob import glob

# Base directory
base_dir = "/data/visitor"

# Pattern to match the directory structure
# We now explicitly look for 'align' or 'beamline' in the subfolder names under RAW_DATA
pattern = os.path.join(
    base_dir,
    "a32*",  # Only folders starting with 'a32'
    "bm32",
    "*",     # date_of_experiment
    "RAW_DATA",
    "*align*",  # Subfolder containing 'align'
    "*",
    "*.h5"
)

# Find all matching .h5 files in 'align' subfolders
align_files = glob(pattern, recursive=True)

# Repeat for 'beamline' subfolders
pattern = os.path.join(
    base_dir,
    "a32*",
    "bm32",
    "*",
    "RAW_DATA",
    "*beamline*",  # Subfolder containing 'beamline'
    "*",
    "*.h5"
)

beamline_files = glob(pattern, recursive=True)

# Combine the results
filtered_files = align_files + beamline_files

# Remove duplicates (if any)
filtered_files = list(set(filtered_files))

# Calculate accumulated size
accumulated_size = 0
file_info = []
for file in filtered_files:
    size = os.path.getsize(file)  # Size in bytes
    accumulated_size += size
    file_info.append((file, size))

# Print individual file sizes (optional)
for file, size in file_info:
    print(f"File: {file}\nSize: {size} bytes\n---")

# Print accumulated size
print(f"\nAccumulated size of all files: {accumulated_size} bytes")
print(f"Accumulated size in MB: {accumulated_size / (1024 * 1024):.2f} MB")
print(f"Accumulated size in GB: {accumulated_size / (1024 * 1024 * 1024):.2f} GB")
```

```python
copied_files = []
import shutil

destination_folder = '/data/visitor/bm32/inhouse/logbookalignBM32'

# Ensure the destination folder exists
os.makedirs(destination_folder, exist_ok=True)

for file in filtered_files:
    try:
        # Copy the file to the destination folder
        shutil.copy2(file, destination_folder)
        size = os.path.getsize(file)
        accumulated_size += size
        copied_files.append((file, size))
        print(f"Copied: {file} (Size: {size} bytes)")
    except Exception as e:
        print(f"Failed to copy {file}: {e}")
```
